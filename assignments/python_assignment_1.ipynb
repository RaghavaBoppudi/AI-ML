## Q1. Movie Ticket Counter
--------------------------------------------------------------
A cinema charges ticket prices based on age:
  - Below 5 years   : Free
  - 5 to 17 years   : Rs. 100
  - 18 to 59 years  : Rs. 200
  - 60 and above    : Rs. 120

Ask the customer for their age and print the ticket price.

Sample run:
  Enter your age: 65  
  Your ticket price is Rs. 120

```python
age = int(input("Enter your age: "))   # int() turns the typed text into a number

# Python runs only the FIRST condition that is true
print(f"Enter your age: {age}")

if age < 5:
    print("Free")
elif age <= 17:
    print("Your ticket price is Rs. 100")
elif age <= 59:
    print("Your ticket price is Rs. 200")
else:
    print("Your ticket price is Rs. 120")
```

## Q2. Electricity Bill Calculator
--------------------------------------------------------------
The electricity board charges as follows:
  - First 100 units       : Rs. 3 per unit
  - Next 100 units (101-200): Rs. 5 per unit
  - Above 200 units       : Rs. 8 per unit

Take the number of units consumed and print the total bill.

Sample run:  
Enter units consumed: 250  
Total bill: Rs. 1200  
(100*3 + 100*5 + 50*8 = 300 + 500 + 400)

```python
units = int(input("Enter units consumed: "))

print(f"Enter units consumed: {units}")

if units <= 100:
    bill = units * 3 # first 100 units at 3 per unit
elif units <= 200:
    bill = 300 + (units - 100) * 5 # 100 units at 3 per unit + next 100 units 5 per unit
else:
    bill = 800 + (units - 200) * 8 # 100 units at 3 per unit + next 100 at 5 per unit + above 200 8 per unit

print("Total bill: Rs.", bill)
```

## Q4. Savings Piggy Bank
--------------------------------------------------------------
Riya puts money into her piggy bank every day. Her goal is
Rs. 1000. Using a while loop, keep asking how much she saved
today and add it to the total. Stop when the total reaches or
crosses Rs. 1000, then print how many days it took.

Sample run:
  Enter today's savings: 300
  Enter today's savings: 450
  Enter today's savings: 400
  Goal reached in 3 days! Total saved: Rs. 1150

```python
total = 0 # initiate the variables for the loop
days = 0

while total < 1000: # repeat until the goal is reached
    daily_savings = int(input("Enter today's savings: ")) # save daily savings in a variable
    total = total + daily_savings # calculate the total
    print(f"Enter today's savings: {daily_savings}")
    days = days + 1

print("Goal reached in", days, "days! Total saved: Rs.", total)
```

## Q5. ATM PIN Check
--------------------------------------------------------------
An ATM's correct PIN is 4321. The user gets only 3 attempts.
  - If the PIN is correct, print "Access granted".
  - If wrong, print "Wrong PIN" and ask again.
  - After 3 wrong attempts, print "Card blocked".

Sample run:
  Enter PIN: 1111
  Wrong PIN
  Enter PIN: 4321
  Access granted

```python
attempts = 0 # initiate the number of attempts

while attempts < 3: # since total number of tries are 3
    pin = int(input("Enter PIN: "))
    if pin == 4321:
        print("Access granted")
        break # stop the loop immediately if the pin is correct
    else:
        print("Wrong PIN")
        attempts = attempts + 1 # increment number of tries on failure

if attempts == 3: # if number of tries reaches 3, block the card
    print("Card blocked")
```

## Q6. Even and Odd Counter
--------------------------------------------------------------
A teacher wants to know how many roll numbers from 1 to N are
even and how many are odd. Ask for N and use a for loop to
count them.

Sample run:  
    Enter N: 15  
    Even numbers: 7  
    Odd numbers: 8

```python
n = int(input("Enter N: ")) 
# initiate the two counters for odd and even
even = 0
odd = 0

for i in range(1, n + 1): # iterate from 1 to n (n inclusive)
    if i % 2 == 0: # even if divisible by 2
        even = even + 1 # incrememnt even counter
    else:
        odd = odd + 1 # increment odd counter
print(f"Enter N: {n}")
print("Even numbers:", even)
print("Odd numbers:", odd)
```

## Q7. Number Guessing Game
--------------------------------------------------------------
The secret number is 27 (store it in a variable). Keep asking
the player to guess until they get it right. After each wrong
guess, tell them "Too high" or "Too low". At the end, print
how many guesses they took.

Sample run:
  Guess the number: 50
  Too high
  Guess the number: 20
  Too low
  Guess the number: 27
  Correct! You took 3 guesses.

```python
secret = 27 # store the secret in a variable "secret" - dumb idea but ok
guesses = 0 # initiate the counter for total guesses
guess = 0

while guess != secret: # keep incrementing as long as the secret isn't guessed
    guess = int(input("Guess the number: ")) # store user input value as a "guess" variable
    guesses = guesses + 1
    if guess > secret: # if guessed number is greater than secret, print "Too high"
        print("Too high")
    elif guess < secret: # if guessed number is lesser than secret, print "Too low"
        print("Too low")

print("Correct! You took", guesses, "guesses.")
```

## Q8. Exam Result with Grade
--------------------------------------------------------------
Take marks (out of 100) of a student and print the grade:
  - 90 and above : A
  - 75 to 89     : B
  - 50 to 74     : C
  - 35 to 49     : D
  - Below 35     : Fail
If the marks entered are less than 0 or more than 100, print
"Invalid marks".

```python
marks = int(input("Enter marks: ")) # store user input in marks variable

if marks < 0 or marks > 100: # make sure the range is between 0 and 100
    print("Invalid marks")
elif marks >= 90:
    print("A")
elif marks >= 75:
    print("B")
elif marks >= 50:
    print("C")
elif marks >= 35:
    print("D")
else:
    print("Fail")
```

## Q9. Sum of Digits on a Bill Number
--------------------------------------------------------------
A shop gives a free gift if the sum of the digits of the bill
number is more than 20. Take the bill number (as an integer)
and use a while loop with % and // to find the sum of its
digits. Print whether the customer gets a gift.

Sample run:
  Enter bill number: 98765
  Sum of digits: 35
  Congratulations! You win a free gift.

```python
num = int(input("Enter bill number: "))
total = 0

while num > 0:
    total = total + num % 10 # % 10 gives the last digit  (98765 -> 5)
    num = num // 10 # // 10 removes the last digit (98765 -> 9876)

print("Sum of digits:", total)
if total > 20:
    print("Congratulations! You win a free gift.")
else:
    print("No gift this time.")
```

## Q10. Staircase Pattern

nested for loops.

  Enter number of steps: 4  
  \*  
  \* \*
