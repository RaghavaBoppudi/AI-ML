# Data Visualization with Pandas  
Python provides 4 ways to plot:  
1. Matplotlib
2. Seaborn
3. Plotly
4. Cufflings

## MatplotLib  

Library to help with plotting 2D visuals.  

Common plot types:
1. Line Plot
2. Scatter Plot
3. Bar Chart
4. Area Plot

### Subplots


### Line Plot
Tracks continuous changes / trends over time.  

### Bar chart  
Compare discrete categories - barh for horizontal  

### Histograms  
View frequency distribution of a single continous variable  

### Stacked bar chart
Compare overall category totals and see how each individual part contributes to that total.  

### Scatter Plot  
Talks about spread and variance in the data - shows relationships and correlations between the two variables  

### Hexbin  
A 2D histogram that helps you visualize the relationship and density of two continuous variables when you have too many data points for a normal scatter plot  

### Pie chart  
Percentages of a whole  

### Andrews Curve  
Helps visualize data with multiple variables on a 2d chart  
Transforms every row into a continous wave - using Fourier series of sine and cosine waves  
Rows with similar measurements produce identical waves.  

### Map visualization  

### Box plot  

Summarizing medians, quartiles, and outliers across groups  

### Area chart  

Visualizes volume or cumulative totals under a curve.  

### Heatmap  

visualizes a 2D table or matrix of numbers as a grid of colored squares, where the color intensity of each cell represents its numerical value.

## Seaborn  

Matplotlib is Python's foundational, low-level plotting library for granular customization, while Seaborn is a high-level wrapper built on top of Matplotlib for quick, attractive statistical charts

## Plotly and Cufflings - 3D visuals

2D visuals like Matplotlib and seaborn usually deliver static images.  

Plotly and Cufflings libraries provide web based, interactive visuals - hover, opan, zoom, toggle, etc.
