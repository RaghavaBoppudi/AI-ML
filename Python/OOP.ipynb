## Special methods: 

#### \_\_str\_\_()

When you say  

    print(person1)

It does not print the name of the person or anything useful. it only says "hey this object exists here in memory.

\_\_str\_\_() can control what gets printed.

It essentially makes the object readable for humans

```python
class Person:

    def __init__(self, name):
        self.name = name

    def __str__(self):
        return f"My name is {self.name}"

person1 = Person("Raghava")

print(person1) ## now since we defined what gets printed using __str__(), we get the statement printed
```


#### \_\_repr\_\_()


```python
class Person:

    def __init__(self, name, surname, year_of_birth):
        self.name = name
        self.surname = surname
        self.year_of_birth = year_of_birth

    def __str__(self):
        return f"{self.name} {self.surname} was born in {self.year_of_birth}."

    def __repr__(self):
        return f"Person(name='{self.name}', surname='{self.surname}', year_of_birth={self.year_of_birth})"


alec = Person("Alec", "Baldwin", 1958)

print(str(alec))
print(repr(alec))
```

## Encapsulation

Essentially means keeping data and the methods that work with that data together, while controlling access to the internal details.

### Internal Attributes
1. Single underscore
2. Double underscore

### Single underscore.

Signals the developers that this is an internal attribute and not to change it.

It does not prevent them from doing that though, it's just a security convention not a guardrail.

```python
class Person:
    def __init__(self, name, surname):
        self._name = name
        self._surname = surname ## using an underscore here signals to other developers that these attributes are internal.

person1 = Person("Raghava", "Boppudi")
```

```python
print(person1._name)
```

### Double underscore.  

Also called name mangling.

```python
class Person:
    def __init__(self, name):
        self.__name = name ## used double underscore here

person1 = Person("Raghava") ## define an object

print(person1.__name) ## returns an error saying person1 does not have an attribute __name.

## BUT

print(person1._Person__name) ## gets printed just fine. Notice how? Because internally __name became __className_name - name mangling
```

## Getter Functions

used to retrieve the value of a private or protected attribute without accessing it directly. 

```python
class Person:
    def __init__(self, name):
        self.__name = name ## use double underscore for abstraction
    
    def get_name(self):
        return self.__name ## define a get_name function that takes itself as an attribute to return its name

person1 = Person("Raghava") ## creating an object

## print(person1.__name) ## throws an error

print(person1.get_name())
```

## Setter function

Used to set the value of a private or protected attribute without accessing it directly.  

You must define a set_name function first

```python
class Person:
    def __init__(self, name, age):
        self.__name = name ## use double underscore for abstraction
        self.__age = age
    
    def get_name(self):
        return self.__name, self.__age

    def set_name(self, new_name, new_age):
        self.__name = new_name ## set a new_name attrribute and assign it to name.
        self.__age = new_age

person1 = Person("Raghava", 25)

print(person1.get_name()) ## prints Raghava because new name isn't set yet

person1.set_name("Aaradhya", 24)

print(person1.get_name())
```

## Inheritance

Child class inherits all of the **methods** of the parent class.

```python
class ChildClass(ParentClass):
    pass
```

We do not need to create another \_\_init\_\_() method inside student - it inherits it from the parent class.

```python
class Person:

    def __init__(self, first_name, last_name, birth_year):
        self._first_name = first_name
        self._last_name = last_name
        self._birth_year = birth_year

    def age(self, current_year):
        return current_year - self._birth_year


class Student(Person):
    pass


rvbpd = Student("Raghava", "Boppudi", 2001)

print(rvbpd._first_name) ## prints without us even defining __init__() inside the child class, nor the other attributes
```

We're reusing functionality from another class instead of writing it again.  

You can alternatively also add additional attributes to the child class. studentID in the below example.

```python
class Student(Person):
    def __init__(self, first_name, last_name, birth_year, studentID):
        self._first_name = first_name
        self._last_name = last_name
        self._birth_year = birth_year
        self._studentID = studentID

rvbpd = Student("Raghava", "Boppudi", 2001, "18951a05b3")

print(rvbpd._first_name) ## this underscore is just the underscore from the abstraction and encapsulation concepts
print(rvbpd._studentID)
```

### Super() function

Instead of repeating all of the parent's \_\_init\_\_() attributes, we can just use super().

```python
class Student(Person):
    def __init__(self, first_name, last_name, birth_year, studentID): ## initialize everything here - parent and child class' attributes
        super().__init__(first_name, last_name, birth_year) ## use super to borow all of the init attributes of the parent class as is.
        self._studentID = studentID ## declare just the unique attributes here

rvbpd = Student("Raghava", "Boppudi", 2001, "18951a05b3")

print(rvbpd._first_name)
print(rvbpd._studentID)
```

## Encapsulation - Composition

Essentially creating sub classes and using them inside parent classes instead of having.  

Break a complicated application into smaller, independent classes

```python
class Engine: ## create a class called Engine
    def start(self): ## define a method called self
        print("Engine is starting...")

class Car: ## create a class called car

    def __init__(self):
        self.engine = Engine() ## create an Engine object and stores it inside the current Car object
        ## Every time we create a Car object, we also create an Engine object inside it

    def start(self):
        print("start engine first")

        self.engine.start() # inside this Car object, find its Engine object, and execute the Engine's start method

        print("engine started, car is starting")

        print("Car is ready to go!")
```

## Encapsulation - Dynamic Extensions  

Adding new attributes or methods to an object or class without changing the original class definition.  

add functionality to an existing object dynamically by wrapping it with other objects

```python
## create a Text class

class Text:

    def render(self): ## create a method called render that returns "Hello"
        return "Hello"

## create a separate boldWrapper class

class BoldWrapper:
    def __init__(self, wrapped):
        self.wrapped = wrapped ## just like creating an object inside the class, but instead accepting an already created object
        
        ## wrapped is a parameter that recieves another object

    def render(self):
        return f"<b>{self.wrapped.render()}</b>" ## add the bold formatting to the wrapped object

class ItalicWrapper:
    def __init__(self, wrapped):
        self.wrapped = wrapped
    
    def render(self):
        return f"<i>(self.wrapped.render())</i>" ## add the italic formatting to the wrapped object

simple = Text() ## create an object called simple for the Text class

bold = BoldWrapper(simple) ## use the bold wrapper on the simple object

italic = ItalicWrapper(simple) ## use the italic wrapper on the simple object

bold_italic = BoldWrapper(italic) ## pass the italic object from above onto the bold wrapper

italic_bold = ItalicWrapper(bold) ## pass the bold object from above onto the italic wrapper

## Print the rendered outputs of all

## remember it is not print(italic_bold), 
## but rather print(italic_bold.render())
# Why? Because print(italic_bold) only prints something like <__main__.ItalicWrapper object at 0x108066fd0>

print(italic_bold.render())
print(bold.render())
print(italic.render())
print(bold_italic.render())
```

### Using inheritance to achieve the same thing rather than wrapping an existing object

```python
class Text:

    def render(self): ## create a method called render that returns "Hello"
        return "Hello"

class BoldText(Text): ## create BoldText as a child class of Text

    def render(self):
        return "<b>" + super().render() + "</b>" 
        ## inherit the render method from Text (superclass) and wrap it in bold

    def __str__(self):
        return self.render()


b = BoldText()

print(b)
```

## Polymorphism and DuckTyping  

### Polymorphism 
Same operation can work with different tyopes of objects, with different behavior based on data type.

```python
def summer(a, b): ## define a function to return the sum of numbers
    return a + b


print(summer(1, int("1"))) ## pass an integer value, and a string value converted to int

print(summer(["a", "b", "c"], ["d", "e"])) ## pass two lists

print(summer("abra", "cadabra")) # pass two strings
```

In the above examples, the behavior of + changed depending on the supplied values

### Ducktyping
A way to support polymorphic behavior.  
The same method can produce two different results based on how and where it is used.  

```python
## create two classes Dog and Cat with the same speak() method

class Dog:

    def speak(self):
        print("Woof!")


class Cat:

    def speak(self):
        print("Meow!")

## create a function to accept an object and execute its speak() method

def make_sound(animal):
    animal.speak()

## now the same speak() method gives out two different outputs based on what object is passed.  

dog = Dog() ## create a dog object in parent class Dog
cat = Cat() ## create a cat object in parent class Cat

make_sound(dog)

make_sound(cat)
```

**Polymorphism** = Same operation, different behavior.  

**Duck Typing** = If an object supports the operation we need, we can use it

## Class Variable & Instance Variable  

Python has two types of variables.  

1. Instance variable  
2. Class variable

```python
class Student:

    def __init__(self, name):
        self.name = name

s1 = Student("Raghav")
s2 = Student("Aaradhya")

## python creates both s1 and s2 and changing one doesn't change the other

print(s1.name)
print(s2.name)

## Because name belongs to each individual object.

## That, is an "Instance variable"
```

```python
## Now lets introduce "Class variable"

class Student:

    school = "FIITJEE"

    def __init__(self, name):
        self.name = name

s1 = Student("Raghav")
s2 = Student("Aaradhya")

## python creates both s1 and s2 and changing one doesn't change the other

print(s1.name)
print(s2.name)

## But since school was declared inside the class and not it's __init__(),
## changing it changes the values below:

print(s1.school)
print(s2.school)

## you don't need to create an object to access a class variable, you can access it directly

print(Student.school)

## One important distinction is the class variable inside the object can still be affected by changing its instance variable, just not on a class level - see below

s1.school = "FIITJEE"
s2.school = "St. Andrews"

print(s1.school)
print(s2.school)

## This is because an instance variable takes precedence over a class variable

print(Student.school)

## notice how the class variable was chaged inside s1 and s2, but the class variable was never actually changed?
## This is Attribute Shadowing
```

**Class variable** is defined inside the Class and outside the \_\_init\_\_() code block.  
**Instance variable** is defined inside the \_\_init\_\_() code block.

### Instance Method

1. Takes self as its first parameter.  
2. Can access the object's instance variables  
3. Can access and modify class variables too

```python
class MyClass:

    class_var = 100 ## class variable

    def __init__(self, value):
        self.value = value ## instance variable

    def show(self): ## this is instance method - see how it takes self as a parameter?
        print(f"Instance value: {self.value}")
        print(f"Class variable before: {MyClass.class_var}")

        self.value += 10 ## take current object's value and add 10
        MyClass.class_var -= 5 ## take class variable and subtract 5

        print(f"Instance value after: {self.value}")
        print(f"Class variable after: {MyClass.class_var}")


obj = MyClass(15) ## set the object value to 15

obj.show()
```

## Class method

    @classmethod  

Operates on class-level information. below example counts the number of objects created

```python
class MyClass:

    count = 0 ## initiate two class variables
    rk = 0

    ## define an init constructor that increments the value of rk everytime it runs
    def __init__(self):
        MyClass.rk += 1

    ## now define a class method  
    ## Instead of receiving self automatically, it receives cls  
    ## cls refers to the current Class (just like how self refers to the current object)  
    @classmethod
    def get_instance_count(cls):
        return cls.rk

## execute the code above

print(MyClass.get_instance_count())
## returns 0 because no objects were created. Create three objects  

a = MyClass()

b = MyClass()

c = MyClass()

print(MyClass.get_instance_count())
```

Can be used for multiple things like:  
1. Count how many objects were created.  
2. Change shared class settings.  
3. Create objects through alternative constructors.

## Static Methods  

    @staticmethod  
Static method does not automatically receive current object or class.  

It behaves just like a **function that lives inside a class**  



```python
class Math:

    @staticmethod
    def add(a, b): ## just a function that adds two numbers
        return a + b


print(Math.add(5, 7)) ## call it directly through the class
```

```python
class MyClass:

    def __init__(self, value):
        self.value = value

    @staticmethod
    def show_value(obj):
        print(obj.value) ## accesses obj.value because we are explicitly passing that value  


a = MyClass(42) ## create an object a for class MyClass with 42 as the attribute value

MyClass.show_value(a) ## here we're passing it
```
